```python
from scipy.stats import spearmanr
import pandas as pd

metricas = ["R2","MAE","MSE","Accuracy","Recall","Precision"]

resultados = []

for m in metricas:
    rho, p = spearmanr(df[m], df["PnL"], nan_policy="omit")
    resultados.append({
        "Métrica": m,
        "Spearman": rho,
        "p-value": p
    })

spearman_df = pd.DataFrame(resultados)
spearman_df = spearman_df.sort_values("Spearman", ascending=False)

print(spearman_df)
```

```python
from scipy.stats import pearsonr

res=[]

for m in metricas:

    p,_ = pearsonr(df[m],df["PnL"])
    s,_ = spearmanr(df[m],df["PnL"],nan_policy="omit")

    res.append({
        "Métrica":m,
        "Pearson":p,
        "Spearman":s
    })

corr_df=pd.DataFrame(res)

corr_long=corr_df.melt(
    id_vars="Métrica",
    var_name="Tipo",
    value_name="Correlación"
)

plt.figure(figsize=(10,5))

sns.barplot(
    data=corr_long,
    x="Métrica",
    y="Correlación",
    hue="Tipo"
)

plt.axhline(0,color="black")
plt.grid(alpha=.3)
plt.tight_layout()
```
